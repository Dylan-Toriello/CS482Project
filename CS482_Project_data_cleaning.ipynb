```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("llm_price_performance_tracker_2026-03-31.csv")
```

```python
# Drop columns chatbot_arena_elo, arena_elo_ci95, arena_votes, elo_benchmark_blend missing 407 out of 453 values
df = df.drop(columns=['chatbot_arena_elo', 'arena_elo_ci95', 'arena_votes', 'elo_benchmark_blend'])
```

```python
# Convert release_year to integer and scrap date to datetime format
df['release_year'] = df['release_year'].astype('Int64')
df['scrape_date'] = pd.to_datetime(df['scrape_date'])
```

```python
# scale raw benchmarks to match the intelligence_score scale of (0 - 100)
raw_benchmarks = ['mmlu_pro', 'gpqa_diamond', 'humanitys_last_exam', 'livecodebench', 'scicode', 'math_500', 'aime_2025']

for benchmark in raw_benchmarks:
    df[benchmark] = df[benchmark] * 100

#fill missing benchmark values with median so we can use in ML model
raw_benchmarks = raw_benchmarks + ['aa_intelligence_index', 'aa_coding_index', 'composite_benchmark']

for benchmark in raw_benchmarks:
    df[benchmark] = df[benchmark].fillna(df[benchmark].median())

```

```python
df.isnull().sum()
```

# 1 Identify top-10 providers by model count and create bar chart on the count of their models.

```python
top10Providers = df['provider'].value_counts().head(10)

# creation of barchart
plt.figure(figsize=(12,6))
top10Providers.plot(kind='bar',color='teal',edgecolor='black')

plt.title('Top 10 providers by Model Count')
plt.xlabel('Provider Name', fontsize=12)
plt.ylabel('Number of Models', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate labels so they don't overlap
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

print(top10Providers)
```

# 2 A box plot of intelligence distribution of different models by pricing tier.

```python
tiers = ['Free', 'Budget', 'Mid', 'Premium', 'Ultra']

plt.figure(figsize=(12,7))
sns.boxplot(x='pricing_tier', y='aa_intelligence_index', data=df, order=tiers,palette='Set2')

plt.title('Intelligence Distribution by Pricing Tier', fontsize=15)
plt.xlabel('Pricing Tier', fontsize=12)
plt.ylabel('Artifical Analysis Intelligence Index', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
